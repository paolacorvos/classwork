```python
# infinite loop - example
# while True:
#    ("Hello")
    
count = 0
while True:
    if count >= 10:
        break # breaks the loop
    count += 1 # count = count + 1
    print(count)
```

```python
# this is a better option for while loop

count = 0

while count < 10:
    count += 1
    print(count)
```

```python
# WHILE LOOP
is_pass_complete: bool = False

while not is_pass_complete:
    password = input("Enter password in only alphabetic characters")
    if not password.isalpha():
        print("Please choose another password with only alphabetic characters")
        continue # restarts the loop from here. Reads nothing below.
    is_pass_complete = True
    print("Password set")
```

```python
# FOR LOOP
from typing import List

nums: List[int] = [1,2,3,4,5]

for num in nums: # num is the temporary variable used to call each item in the collection in the code block
    num +=1
    if num % 2 != 0: # if odd
        # break would break the loop
        continue # do not print anything odd
    print(num)
```

```python
# NESTED LOOP
sentence: str = "This is a Python course."

words: List[str] = sentence.split(" ")
used_vowels = []

for word in words:
    vowels = ["a", "e", "i", "o", "u", "y"]
    for letter in word:
        if letter in vowels:
            used_vowels.append(letter)

    
print(used_vowels)
```

```python
# what are functions
# functions take an input (argument), processes it, and returns an output (return)

def shorten_names(names_list: List[str], chars: int = 3) -> List[str]:
   
    """ 
    Returns a list of strings that are the first three 
    characters of the word wiht a period apended.
    
    Args:
        names_list (List[str]): A list of strings that will be shortened.
        chars (int):The number of characters to shorten each
        string to. Must be greater than 0.
    Return:
        Lis[str]: A list shortened strings to 3 characters
    """
    if type(chars) != int:
        raise TypeError("Chars must be an int")
    if chars < 0:
        raise ValueError("Chars must be greater than 0")
    names_list = [f"{name[:chars]}." for name in names_list]
    return names_list
```

```python
bar_names = ["Ottoman Empire", "France", "Spain"]
shorten_names(bar_names, chars=4)
```
