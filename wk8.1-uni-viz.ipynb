```python
# %conda install seaborn
# %conda install matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/soccer-players.csv"
df = pd.read_csv(url)
df.head()
```

```python
import scipy.stats as stats
```

```python
# Anatomy of a Viz
plt.figure(figsize=(8,5))
hts = df["height_in_cm"]
bins = round(hts.count() ** (1/3))
plt.hist(x=hts, bins=bins, edgecolor="black", color="pink")
plt.xlabel("Height in CM")
plt.ylabel("Frequency")
plt.title("Histogram of Heights (in cm) of Soccer Players")
plt.grid(True)
sk = stats.skew(hts, nan_policy="omit")
kurt = stats.kurtosis(hts, nan_policy="omit")
text = f"Skew: {sk:.3f}\nKurtosis: {kurt:.3f}"
plt.text(
    x=0.05, 
    y=.90, 
    s=text, 
    # ha="center", 
    va="center", 
    fontsize=12, 
    color="hotpink",
    transform=plt.gca().transAxes)
plt.axvline(x=hts.mean(), color="hotpink", linestyle="--", label="Mean")
plt.axvline(x=hts.median(), color="cyan", linestyle="--", label="Median")
plt.legend()
plt.show()
```

```python
# Histogram in SNS
df1 = df.loc[df["height_in_cm"]>150]
plt.figure(figsize=(8,5))
sns.set_theme(style="darkgrid")
hist = sns.histplot(data=df1, x="height_in_cm", kde=True)
plt.xlabel("Height (cm)")
plt.title("Histogram of Heights (in cm) of Soccer Players")
hts = df1["height_in_cm"]
sk = stats.skew(hts, nan_policy="omit")
kurt = stats.kurtosis(hts, nan_policy="omit")
text = f"Skew: {sk:.3f}\nKurtosis: {kurt:.3f}"
plt.text(
    x=0.05, 
    y=.90, 
    s=text, 
    # ha="center", 
    va="center", 
    fontsize=12, 
    color="hotpink",
    transform=plt.gca().transAxes)
plt.axvline(x=hts.mean(), color="hotpink", linestyle="--", label="Mean")
plt.axvline(x=hts.median(), color="cyan", linestyle="--", label="Median")
plt.legend()
plt.show()
```
